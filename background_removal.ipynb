```python
from PIL import Image
from pathlib import Path
from tqdm.notebook import tqdm
from rembg import remove, new_session
import pandas as pd
```

```python
# Training set

# Load data

DIR = Path("./aml-feathers/data")

birds_train_df = pd.read_csv(DIR / "train_images.csv")

birds_train_df.head()

# Preprocessing

DIR_working = Path("./aml-feathers/data")
output_map = Path("train_images_rembg")
output_path = DIR_working / output_map

# Create output map
try:
    output_path.mkdir(exist_ok=True, parents=True)
    print(f"Map '{output_map}' succesfully created")
except Exception as e:
    print(f"Error while creating map: {e}")

session = new_session() 

for img_path in tqdm(birds_train_df["image_path"], desc = "Background removal"):
    input = Image.open(DIR / img_path.lstrip("/"))
    output = remove(input, session=session)
    img_name = img_path.lstrip("/train_images/").replace(".jpg", ".png")
    output_img_path = output_path / img_name
    output.save(output_img_path)
```
