We have a vector of 312 attributes per class. Lets see if we can reduce it's dimensionality using PCA


```python
import numpy as np
attrs = np.load("../data/attributes.npy", allow_pickle=True)
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import umap

```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Replace with your data
class_names = [i + 1 for i in range(200)]

scaler = StandardScaler()
attrs_scaled = scaler.fit_transform(attrs)

pca = PCA(n_components=65)
pca_2d = pca.fit_transform(attrs_scaled)

print(f'Cumulative variance: {np.sum(pca.explained_variance_ratio_):.1%}')
```

```python
import pandas as pd
df = pd.DataFrame({
    'className': class_names,
    'pca': list(pca_2d)
})
```
